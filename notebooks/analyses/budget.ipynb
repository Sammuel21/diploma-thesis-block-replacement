## Budget Estimation Strategies

This notebook explores the selection strategies for assignment of block-level budgets across LLM.
First ideas are defined in `/analyses/importance.ipynb`.

### Budget Allocation Ideas

#### 1. Operator Sparsity %

Let original operator $f_{\ell_{MLP}}$ have parameter count $B$ (budget). New parameter count of replacement operator $g_{\ell_{MLP}}$ will be $xB$ where $x$ is the target sparsity ratio. This way we can allocate block-level operator sparsity.

- important: $x$ defines only total parameter count of new replacement operator
- important: does not account for degree of non-linearity the operator requires

##### Optimal sparsity allocation (operator)

We can use different importance estimation strategies to determine the optimal parameter count. For example a good baseline would be to utilize the BI/MLP(residual)-score to weight the total parameter budget across blocks, corresponding block weight $w_{\ell}$ would serve as operator sparsity ratio.
