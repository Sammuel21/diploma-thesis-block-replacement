### Budget Allocation Ideas

#### 1. Operator Sparsity %

Let original operator $f_{\ell_{MLP}}$ have parameter count $B$ (budget). New parameter count of replacement operator $g_{\ell_{MLP}}$ will be $xB$ where $x$ is the target sparsity ratio. This way we can allocate block-level operator sparsity.

- important: $x$ defines only total parameter count of new replacement operator
- important: does not account for degree of non-linearity the operator requires

#### 1. Hybrid linear-nonlinear replacement (future work)

A hybrid replacement may combine a linear branch and a nonlinear correction:

$$
F_\ell(x)=L_\ell(x)+G_\ell(x).
$$

A dense $L_\ell(x)=A_\ell x$ has fixed $d_{\mathrm{model}}\times d_{\mathrm{model}}$ dimensions. Configurable linear capacity therefore requires a factorized branch $U_\ell(V_\ell x)$ with adjustable rank. The nonlinear branch can use an adjustable hidden width, for example in a compact MLP or SwiGLU.

A possible future extension would divide an assigned block budget $B_\ell$ between the linear rank and nonlinear width. This is not yet a defined allocation method: the budget variable, feasible integer projection, branch families, comparison points, and rule for selecting the split remain open. BI or another importance score may determine the total block budget, but it does not by itself determine the linear/nonlinear split.

**Status.** This is a researcher-proposed operator direction and open experiment, not an established method or empirical finding. The equation is a project-proposed operator definition and requires no external citation.
